```python
import numpy as np
import os

def analyze_neuralhash_dat(file_path: str):
    """Correctly analyzes NeuralHash .dat file with header handling"""
    try:
        # Read entire file as bytes
        with open(file_path, 'rb') as f:
            raw_bytes = f.read()
        
        # Extract header and matrix data
        header = raw_bytes[:128]  # First 128 bytes are header
        matrix_bytes = raw_bytes[128:128 + 128*96*4]  # Next 49,152 bytes are matrix
        
        # Convert to float32 matrix
        M = np.frombuffer(matrix_bytes, dtype=np.float32).reshape(128, 96)
        
        return {
            "file_size": len(raw_bytes),
            "header_size": len(header),
            "matrix_size": len(matrix_bytes),
            "shape": M.shape,
            "global_mean": float(np.mean(M)),
            "global_std": float(np.std(M)),
            "min_value": float(np.min(M)),
            "max_value": float(np.max(M)),
            "sample_values": M[:3, :3].tolist(),  # 3x3 sample
        }
    
    except Exception as e:
        return {"error": str(e), "type": type(e).__name__}

# Usage
file_path = r'D:\FYP\NeuralHASH\AppleNeuralHashAlgorithm\model\neuralhash_128x96_seed1.dat'
stats = analyze_neuralhash_dat(file_path)
print(stats)
```

```python
import numpy as np
import os

def extract_neuralhash_matrix(file_path: str):
    """Extracts the full 128x96 matrix from NeuralHash .dat file"""
    try:
        # Read entire file as bytes
        with open(file_path, 'rb') as f:
            raw_bytes = f.read()
        
        # Extract header (128 bytes) and matrix data (128*96*4 = 49,152 bytes)
        header = raw_bytes[:128]
        matrix_bytes = raw_bytes[128:128 + 128*96*4]
        
        # Convert to float32 matrix and return as nested lists
        M = np.frombuffer(matrix_bytes, dtype=np.float32).reshape(128, 96)
        return M.tolist()  # Convert to Python list of lists
    
    except Exception as e:
        return {"error": str(e), "type": type(e).__name__}

# Usage - Replace with your actual file path
file_path = r'D:\FYP\NeuralHASH\AppleNeuralHashAlgorithm\model\neuralhash_128x96_seed1.dat'
full_matrix = extract_neuralhash_matrix(file_path)

# To save the matrix to a file (recommended due to large size)
with open('neuralhash_matrix.txt', 'w') as f:
    for row in full_matrix:
        f.write(','.join(map(str, row)) + '\n')

print(f"Matrix shape: {len(full_matrix)}x{len(full_matrix[0])}")
print("First row sample:", full_matrix[0][:5])
```
